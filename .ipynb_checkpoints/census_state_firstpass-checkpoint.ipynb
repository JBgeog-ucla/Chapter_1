```python
#The following code is from ChatGPT
#not using it b/c doesn't make clear Kern as standout.

# 1. Setup the Grid
# Calculate rows/cols based on how many counties you have
num_counties = len(counties)
cols = 3
rows = math.ceil(num_counties / cols)

fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows), dpi=300, sharex=True)
fig.suptitle('Post-WWII Average Farm Size: County vs. State', fontsize=20, weight='bold', y=1.02)

# Flatten axes array for easy looping
axes_flat = axes.flatten()

# 2. Loop through counties and plot
for i, county in enumerate(counties):
    ax = axes_flat[i]
    
    # Plot State of CA as a background reference on every facet
    pivoted_data['Avg. Farm Size', 'State of CA'].plot(
        ax=ax, color='grey', linestyle='--', alpha=0.5, label='State'
    )
    
    # Plot the specific County
    pivoted_data['Avg. Farm Size', county].plot(
        ax=ax, color='tab:blue', marker='o', markersize=4, linewidth=2, label=county
    )
    
    # Formatting each facet
    ax.set_title(f'{county} County', fontsize=14, fontweight='bold')
    ax.grid(True, alpha=0.3)
    ax.yaxis.set_major_formatter(mtick.StrMethodFormatter('{x:,.0f}'))
    
    if i % cols == 0:
        ax.set_ylabel('Acres')

# 3. Clean up empty subplots
for j in range(i + 1, len(axes_flat)):
    fig.delaxes(axes_flat[j])

plt.tight_layout()
plt.show()
fig.savefig('Average Farm Size facets', bbox_inches='tight', dpi=300)
```

```python
#Setup the Grid
num_counties = len(counties)
cols = 3
rows = math.ceil(num_counties / cols)

#sharey=True forces all plots to the same scale
#Width of the entire figure = 18 inches
#5 rows= height of fig
fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows), sharex=True, sharey=True)
fig.suptitle('Post-WWII Average Farm Size: County vs. State', fontsize=20, weight='bold', y=1.02)

#flattens 2-D grid into a 1d list
#ex: ax[0], ax[1], ...
axes_flat = axes.flatten()

#Loop using zip to avoid manual indexing (i)
#state data gets added to every axis
state_ref = pivoted_data['Avg. Farm Size', 'State of CA']

#walks through two lists at once
#zip creates a series of pairs
#eg: (Fresno, Box 1), (Kern, Box 2), etc...
for county, ax in zip(counties, axes_flat):
    # Plot State
    ax.plot(state_ref, color='k', ls='--', alpha=0.8, label='State')
    # Plot County
    ax.plot(pivoted_data['Avg. Farm Size', county], marker='o', lw=2, label=county, color= 'olivedrab')
    
    ax.set_title(f'{county} County', fontweight='bold')
    ax.grid(True, alpha=0.3)
    #ax.set_ylabel('Acres', fontsize=15,)
    #ax.set_xlabel('Year', fontsize=15,)

#y-axis label
fig.text(0.02, 0.5, 'Acres', va='center', rotation='vertical', fontsize=16, weight='bold')
#x-axis label 
fig.text(0.5, 0.01, 'Year', va='center', fontsize=16, weight='bold')

#adding custom legends
custom_lines = [Line2D([0], [0], color='olivedrab', lw=2),
                Line2D([0], [0], color='k',ls='--', lw=2)]

fig.legend(custom_lines, ['County', 'State'],
             loc='lower right', 
          bbox_to_anchor=(0.75, 0.3), # Coordinates: 0.8 is to the right of the facet
          borderaxespad=0.)

#Remove extra plot in SE corner
for ax in axes_flat[len(counties):]:
    fig.delaxes(ax)
plt.tight_layout()

#nudges plots over to make room for Acres text
plt.subplots_adjust(left=0.07, bottom=0.05)

fig.savefig('Average Farm Size facets same scale', bbox_inches='tight', dpi=300)
```

```python
#Setup the Grid
num_counties = len(counties)
cols = 3
rows = math.ceil(num_counties / cols)

#sharey=True forces all plots to the same scale
#Width of the entire figure = 18 inches
#5 rows= height of fig
fig, axes = plt.subplots(rows, cols, figsize=(18, 5 * rows), sharex=True, sharey=True)
fig.suptitle('Post-WWII Harvested Cropland: San Joaquin Valley', fontsize=20, weight='bold', y=1.02)

#flattens 2-D grid into a 1d list
#ex: ax[0], ax[1], ...
axes_flat = axes.flatten()

#Loop using zip to avoid manual indexing (i)
#state data gets added to every axis
#state_ref = pivoted_data['Cropland harvested  (acres)', 'State of CA']

#walks through two lists at once
#zip creates a series of pairs
#eg: (Fresno, Box 1), (Kern, Box 2), etc...
for county, ax in zip(counties, axes_flat):
    # Plot State
    #ax.plot(state_ref, color='k', ls='--', alpha=0.8, label='State')
    # Plot County
    ax.plot(pivoted_data['Cropland harvested  (acres)', county], marker='o', lw=2, label=county, color= 'maroon')
    ax.plot(pivoted_data['Cropland (1000 acres and over: acres)', county], marker='o', lw=2, color='darkseagreen')
    ax.annotate("1950: Data Not Available", xy=(0.05, 0.9), xycoords='axes fraction', color='darkseagreen')
    
    ax.set_title(f'{county} County', fontweight='bold')
    ax.grid(True, alpha=0.3)
    #ax.set_ylabel('Acres', fontsize=15,)
    #ax.set_xlabel('Year', fontsize=15,)

#adding custom legends
custom_lines = [Line2D([0], [0], color='darkseagreen', lw=2),
                Line2D([0], [0], color='maroon', lw=2)]

fig.legend(custom_lines, ['1000 Acres and Over', 'Harvested Cropland'],
             loc='lower right', 
          bbox_to_anchor=(0.8, 0.3), # Coordinates: 1.05 is to the right of the facet
          borderaxespad=0.)

#y-axis label
fig.text(0.02, 0.5, 'Acres', va='center', rotation='vertical', fontsize=16, weight='bold')
#x-axis label 
fig.text(0.5, 0.01, 'Year', va='center', fontsize=16, weight='bold')

#Remove extra plot in SE corner
for ax in axes_flat[len(counties):]:
    fig.delaxes(ax)

#plt.ticklabel_format(style='plain')    # to prevent scientific notation.
# after plotting the data, format the labels
current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels(['{:,.0f}'.format(x) for x in current_values])

plt.tight_layout()

#nudges plots over to make room for Acres text
plt.subplots_adjust(left=0.08, bottom=0.05)

fig.savefig('Harvested Cropland facets same scale', bbox_inches='tight', dpi=300)
```
